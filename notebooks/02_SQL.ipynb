```python
import pandas as pd
import sqlite3

# Load cleaned data (suppress mixed type warning)
df = pd.read_csv('../data/claims_cleaned.csv', low_memory=False)

conn = sqlite3.connect('../data/fraud_detection.db')
cursor = conn.cursor()

# --- Table 1: providers ---
cursor.execute('''
    CREATE TABLE IF NOT EXISTS providers (
        Provider        TEXT PRIMARY KEY,
        PotentialFraud  TEXT,
        FraudLabel      INTEGER
    )
''')

# --- Table 2: beneficiaries ---
cursor.execute('''
    CREATE TABLE IF NOT EXISTS beneficiaries (
        BeneID          TEXT PRIMARY KEY,
        DOB             TEXT,
        DOD             TEXT,
        Gender          INTEGER,
        Race            INTEGER,
        State           INTEGER,
        County          INTEGER,
        PatientDied     INTEGER,
        PatientAge      INTEGER
    )
''')

# --- Table 3: claims ---
cursor.execute('''
    CREATE TABLE IF NOT EXISTS claims (
        ClaimID                 TEXT PRIMARY KEY,
        BeneID                  TEXT,
        Provider                TEXT,
        ClaimType               TEXT,
        ClaimStartDt            TEXT,
        ClaimEndDt              TEXT,
        ClaimDuration           REAL,
        HospitalStayDays        REAL,
        InscClaimAmtReimbursed  REAL,
        DeductibleAmtPaid       REAL,
        AttendingPhysician      TEXT,
        OperatingPhysician      TEXT,
        OtherPhysician          TEXT,
        FraudLabel              INTEGER
    )
''')

conn.commit()
print("Tables created")
```
