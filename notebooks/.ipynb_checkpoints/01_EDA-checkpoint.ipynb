```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Loading all 4 files
train_beneficiary = pd.read_csv('../data/Train_Beneficiarydata-1542865627584.csv')
train_inpatient   = pd.read_csv('../data/Train_Inpatientdata-1542865627584.csv')
train_outpatient  = pd.read_csv('../data/Train_Outpatientdata-1542865627584.csv')
train_labels      = pd.read_csv('../data/Train-1542865627584.csv')

print("Beneficiary:", train_beneficiary.shape)
print("Inpatient:  ", train_inpatient.shape)
print("Outpatient: ", train_outpatient.shape)
print("Labels:     ", train_labels.shape)
```

```python
# inpatient vs outpatient
train_inpatient['ClaimType'] = 'Inpatient'
train_outpatient['ClaimType'] = 'Outpatient'

# Combine claims
all_claims = pd.concat([train_inpatient, train_outpatient], ignore_index=True)

# Merging with beneficiary info
df = all_claims.merge(train_beneficiary, on='BeneID', how='left')

# Merging fraud labels
df = df.merge(train_labels, on='Provider', how='left')

# Encoding fraud label
df['FraudLabel'] = (df['PotentialFraud'] == 'Yes').astype(int)

print("Final shape:", df.shape)
print("Fraud rate: ", round(df['FraudLabel'].mean() * 100, 2), "%")
```

```python
# Missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0].sort_values('Missing %', ascending=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Fraud vs non-fraud count
df['PotentialFraud'].value_counts().plot(
    kind='bar', ax=axes[0], color=['#4C9BE8','#E85C5C'], edgecolor='white'
)
axes[0].set_title('Fraud vs Non-Fraud Providers')
axes[0].set_xlabel('')
axes[0].set_ylabel('Count')
axes[0].tick_params(rotation=0)

# Claim type breakdown
df.groupby(['ClaimType','PotentialFraud']).size().unstack().plot(
    kind='bar', ax=axes[1], color=['#4C9BE8','#E85C5C'], edgecolor='white'
)
axes[1].set_title('Fraud by Claim Type')
axes[1].set_xlabel('')
axes[1].tick_params(rotation=0)

plt.tight_layout()
plt.savefig('../outputs/fraud_distribution.png', dpi=150)
plt.show()
```

```python
# Fill missing reimbursement values with 0
df['InscClaimAmtReimbursed'] = df['InscClaimAmtReimbursed'].fillna(0)
df['DeductibleAmtPaid'] = df['DeductibleAmtPaid'].fillna(0)

# Compare reimbursement amounts: fraud vs legit
plt.figure(figsize=(10, 4))
for label, color in zip([0, 1], ['#4C9BE8', '#E85C5C']):
    subset = df[df['FraudLabel'] == label]['InscClaimAmtReimbursed']
    subset.clip(upper=subset.quantile(0.95)).plot(
        kind='hist', bins=50, alpha=0.6, color=color,
        label='Fraud' if label == 1 else 'Legit'
    )
plt.title('Claim Reimbursement Amount: Fraud vs Legit')
plt.xlabel('Amount ($)')
plt.legend()
plt.tight_layout()
plt.savefig('../outputs/reimbursement_distribution.png', dpi=150)
plt.show()
```

```python
# Drop columns with >95% missing OR all-same value
cols_to_drop = [
    'ClmProcedureCode_4', 'ClmProcedureCode_5', 'ClmProcedureCode_6',
    'ClmProcedureCode_3', 'ClmProcedureCode_2', 'ClmProcedureCode_1',
    'ClmDiagnosisCode_10', 'ClmDiagnosisCode_9', 'ClmDiagnosisCode_8',
    'ClmDiagnosisCode_7', 'ClmDiagnosisCode_6', 'ClmDiagnosisCode_5',
    'ClmDiagnosisCode_4', 'ClmDiagnosisCode_3',
]

df.drop(columns=cols_to_drop, inplace=True)
print("Remaining columns:", df.shape[1])
```

```python
date_cols = ['ClaimStartDt', 'ClaimEndDt', 'AdmissionDt', 'DischargeDt', 'DOB', 'DOD']
for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')

# Claim duration in days
df['ClaimDuration'] = (df['ClaimEndDt'] - df['ClaimStartDt']).dt.days

# Hospital stay duration (inpatient only)
df['HospitalStayDays'] = (df['DischargeDt'] - df['AdmissionDt']).dt.days.fillna(0)

# Patient age at claim
df['PatientAge'] = (df['ClaimStartDt'] - df['DOB']).dt.days // 365

print("Claim duration sample:", df['ClaimDuration'].describe())
print("Patient age sample:", df['PatientAge'].describe())
```

```python
# Physicians: missing = no physician involved, fill with 'NONE'
for col in ['AttendingPhysician', 'OperatingPhysician', 'OtherPhysician']:
    df[col] = df[col].fillna('NONE')

# Diagnosis codes: fill with 'UNKNOWN'
for col in ['ClmDiagnosisCode_1', 'ClmDiagnosisCode_2', 'ClmAdmitDiagnosisCode']:
    df[col] = df[col].fillna('UNKNOWN')

# Numeric fills
df['DeductibleAmtPaid'] = df['DeductibleAmtPaid'].fillna(0)
df['HospitalStayDays']  = df['HospitalStayDays'].fillna(0)

# Flag: did patient die?
df['PatientDied'] = df['DOD'].notna().astype(int)

print("Missing values remaining:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Avg claim amount by fraud
df.groupby('PotentialFraud')['InscClaimAmtReimbursed'].mean().plot(
    kind='bar', ax=axes[0], color=['#4C9BE8','#E85C5C'], edgecolor='white'
)
axes[0].set_title('Avg Reimbursement: Fraud vs Legit')
axes[0].set_ylabel('Amount ($)')
axes[0].tick_params(rotation=0)

# Avg claim duration by fraud
df.groupby('PotentialFraud')['ClaimDuration'].mean().plot(
    kind='bar', ax=axes[1], color=['#4C9BE8','#E85C5C'], edgecolor='white'
)
axes[1].set_title('Avg Claim Duration: Fraud vs Legit')
axes[1].set_ylabel('Days')
axes[1].tick_params(rotation=0)

# Avg patient age by fraud
df.groupby('PotentialFraud')['PatientAge'].mean().plot(
    kind='bar', ax=axes[2], color=['#4C9BE8','#E85C5C'], edgecolor='white'
)
axes[2].set_title('Avg Patient Age: Fraud vs Legit')
axes[2].set_ylabel('Age')
axes[2].tick_params(rotation=0)

plt.tight_layout()
plt.savefig('../outputs/fraud_insights.png', dpi=150)
plt.show()
```

```python
df.to_csv('../data/claims_cleaned.csv', index=False)
print("Saved claims_cleaned.csv —", df.shape)
```
